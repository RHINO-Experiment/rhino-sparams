# Notebook For Visualising the Antenna $S_{11}$

```python
import numpy as np
import matplotlib.pyplot as plt

```

```python
class simSparams:
    def __init__(self,
                filepath,
                label):
        self.freq_ghz, self.s11_mag_db = self.read_in_from_txt(filepath)
        self.freqs_mhz = self.freq_ghz * 1000
        self.label = label
        pass
    def read_in_from_txt(self,
                         filepath):
        data = np.loadtxt(filepath,
                           dtype=float,
                           comments='#')
        freq_ghz = data[:,0]
        s11_mag_db = data[:,1]
        return freq_ghz, s11_mag_db
```

```python
short_horn = simSparams('short_horn/RhinoS11_OneSection.txt', label='One Flare Section Horn')
```

```python
plt.plot(short_horn.freqs_mhz, short_horn.s11_mag_db, label=short_horn.label)
plt.ylabel('$|S_{11}|$ [dB]')
plt.xlabel('$\nu$ [MHz]')
plt.gcf().set_size_inches((10, 6))
plt.legend()
plt.show()
```
